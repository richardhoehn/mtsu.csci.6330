# Monte Carlo π — Point Visualization

Load a CSV produced by `Pi-Estimator.c` and color the points based on whether they are inside the unit circle.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Change this to your CSV filename or path
filename = "results_100000000.csv"

df = pd.read_csv(filename)
df.head()
```

```python
inside = df["inside"].astype(str).str.lower().eq("true")

fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(df.loc[~inside, "x"], df.loc[~inside, "y"], s=2, c="tomato", label="Outside (false)", alpha=0.6, rasterized=True)
#ax.scatter(df.loc[inside, "x"], df.loc[inside, "y"], s=2, c="seagreen", label="Inside (true)", alpha=0.6, rasterized=True)

ax.set_aspect("equal", adjustable="box")
ax.set_xlim(-1.05, 1.05)
ax.set_ylim(-1.05, 1.05)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title(f"π Estimation — {len(df):,} Points")
ax.legend(markerscale=4)
ax.grid(alpha=0.15)
plt.show()

print(f"Tests: {len(df):,}")
print(f"Inside: {inside.sum():,}")
print(f"Estimated π: {4 * inside.mean():.8f}")
```
